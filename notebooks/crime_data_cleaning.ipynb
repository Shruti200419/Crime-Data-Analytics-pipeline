```python
import pandas as pd
import numpy as np

df=pd.read_csv(r"C:\Users\HONOR\Downloads\MR data - Compiled Data Set.csv")
df.head()
```

```python
df.info()
```

```python
df.columns= df.columns.str.strip()
df.columns= ['city','title','text','murder_reason','murder_child_victims','murder_male_adult_victims','murder_female_adult_victims',
              'kidnap_child_victims', 'kidnap_male_adult_victims', 'kidnap_female_adult_victims',
              'crime_against_women_type', 'total_adult_victims', 'total_child_victims']
```

```python
numeric_cols=['murder_child_victims','murder_male_adult_victims','murder_female_adult_victims',
                 'kidnap_child_victims', 'kidnap_male_adult_victims', 'kidnap_female_adult_victims',
                 'total_adult_victims', 'total_child_victims']
df[numeric_cols] = df[numeric_cols].fillna(0).astype(int)
category_cols = ['murder_reason','crime_against_women_type']
df[category_cols]= df[category_cols].fillna('Not Applicable')
df.isnull().sum()
```

```python
df['city'].unique()
```

```python
df['murder_reason'].unique()
```

```python
df['murder_reason'] = df['murder_reason'].replace({'Unknown reasons': 'Unknown/Other', 
                                                   'Property Dispute': 'Property/Land Disputes'})
df['murder_reason'].unique()
```

```python
df['crime_against_women_type'].unique()
```

```python
df['crime_against_women_type'] = df['crime_against_women_type'].str.replace(' ', '', regex=False)
df['crime_against_women_type'].unique()
```

```python
df[df['crime_against_women_type'] == '11'][['city', 'title', 'text']]
```

```python
df['crime_against_women_type'] = df['crime_against_women_type'].replace('11', 'Other/Undocumented')
df['crime_against_women_type'].unique()
```

```python
df.duplicated().sum()
```

```python
df['calculated_adult_total'] = df['murder_male_adult_victims'] + df['murder_female_adult_victims'] + df['kidnap_male_adult_victims'] + df['kidnap_female_adult_victims']

mismatch_adult = df[df['total_adult_victims'] != df['calculated_adult_total']]
mismatch_adult[['title', 'total_adult_victims', 'calculated_adult_total']]
```

```python
mismatch_adult['has_caw_code']= mismatch_adult.index.map(lambda i: df.loc[i, 'crime_against_women_type'] != 'Not Applicable')
mismatch_adult['has_caw_code'].value_counts()
```

```python
df = df.drop(columns=['calculated_adult_total'])

# Sanity check: no negative values anywhere in numeric columns
numeric_cols = ['murder_child_victims', 'murder_male_adult_victims', 'murder_female_adult_victims',
                 'kidnap_child_victims', 'kidnap_male_adult_victims', 'kidnap_female_adult_victims',
                 'total_adult_victims', 'total_child_victims']

(df[numeric_cols] < 0).sum()
```

```python
df['total_victims'] = df['murder_child_victims'] + df['murder_male_adult_victims'] + df['murder_female_adult_victims'] + df['kidnap_child_victims'] + df['kidnap_male_adult_victims'] + df['kidnap_female_adult_victims'] + df['total_adult_victims'] + df['total_child_victims']

df['crime_category'] = np.select(
    [
        df['murder_reason'] != 'Not Applicable',
        df['crime_against_women_type'] != 'Not Applicable',
        (df['kidnap_child_victims'] + df['kidnap_male_adult_victims'] + df['kidnap_female_adult_victims']) > 0
    ],
    ['Murder', 'Crime Against Women', 'Kidnapping'],
    default='Unclassified'
)

df['crime_category'].value_counts()
```

```python
df = df[['city', 'title', 'text', 'crime_category', 'total_victims',
         'murder_reason', 'murder_child_victims', 'murder_male_adult_victims', 'murder_female_adult_victims',
         'kidnap_child_victims', 'kidnap_male_adult_victims', 'kidnap_female_adult_victims',
         'crime_against_women_type', 'total_adult_victims', 'total_child_victims']]

df.info()
df.head()
```

```python
df['crime_against_women_type'] = df['crime_against_women_type'].replace('NotApplicable', 'Not Applicable')

df['crime_category'] = np.select(
    [
        df['murder_reason'] != 'Not Applicable',
        df['crime_against_women_type'] != 'Not Applicable',
        (df['kidnap_child_victims'] + df['kidnap_male_adult_victims'] + df['kidnap_female_adult_victims']) > 0
    ],
    ['Murder', 'Crime Against Women', 'Kidnapping'],
    default='Unclassified'
)

df['crime_category'].value_counts()
```

```python
df.to_csv('crime_data_clean.csv', index=False)
print("Saved:",df.shape)
```
